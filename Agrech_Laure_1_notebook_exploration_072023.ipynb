```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Import the datasets
customers = pd.read_csv('olist_customers_dataset.csv')
geolocation = pd.read_csv('olist_geolocation_dataset.csv')
items = pd.read_csv('olist_order_items_dataset.csv')
payments = pd.read_csv('olist_order_payments_dataset.csv')
reviews = pd.read_csv('olist_order_reviews_dataset.csv')
orders = pd.read_csv('olist_orders_dataset.csv')
products = pd.read_csv('olist_products_dataset.csv')
sellers = pd.read_csv('olist_sellers_dataset.csv')
translations = pd.read_csv("product_category_name_translation.csv")

# Display the first few rows of each dataset
print('Customers:')
display(customers.head())
print('\nGeolocation:')
display(geolocation.head())
print('\nOrder Items:')
display(items.head())
print('\nOrder Payments:')
display(payments.head())
print('\nOrder Reviews:')
display(reviews.head())
print('\nOrders:')
display(orders.head())
print('\nProducts:')
display(products.head())
print('\nSellers:')
display(sellers.head())
print('\nTranslations:')
display(translations.head())
```

```python
# Create the dictionary of datasets
datasets_dict = {
    "customers": customers,
    "geolocation": geolocation,
    "items": items,
    "payments": payments,
    "reviews": reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "traductions": traductions
}

data = []

for name, dataset in datasets_dict.items():
    row_dict = {
        'dataset': name,
        'rows': dataset.shape[0],
        'columns': dataset.shape[1],
        'missing_values': dataset.isnull().sum().sum(),
        'duplicated_values': dataset.duplicated().sum(),
    }
    data.append(row_dict)

summary = pd.DataFrame(data)
summary.set_index('dataset', inplace=True)

display(summary)
```

```python
datasets = [customers, geolocation, items, payments, reviews, orders, products, sellers]
dataset_names = ['Customers', 'Geolocation', 'Order Items', 'Order Payments', 'Order Reviews', 'Orders', 'Products', 'Sellers']

for dataset, name in zip(datasets, dataset_names):
    print(f'\n{name}:')
    
    # Create Structure DataFrame
    structure_df = pd.DataFrame({
        'Columns': dataset.columns,
        'Data type': [dataset[col].dtype for col in dataset.columns]
    }).set_index('Columns')
    
    # Create Count DataFrame
    count_df = pd.DataFrame({
        'Columns': dataset.columns,
        'Count': [dataset[col].count() for col in dataset.columns]
    }).set_index('Columns')
    
    # Create Missing Values DataFrame
    missing_values_df = pd.DataFrame({
        'Columns': dataset.columns,
        'Missing values': [dataset[col].isnull().sum() for col in dataset.columns]
    }).set_index('Columns')
    
    # Create Unique Values DataFrame
    unique_values_df = pd.DataFrame({
        'Columns': dataset.columns,
        'Unique values': [dataset[col].nunique() for col in dataset.columns]
    }).set_index('Columns')
    
    # Create Duplicates DataFrame
    duplicates_df = pd.DataFrame({
        'Columns': dataset.columns,
        'Duplicates': [dataset[col].duplicated().sum() for col in dataset.columns]
    }).set_index('Columns')
    
    # Concatenate all the DataFrames side by side
    summary_df = pd.concat([structure_df, count_df, unique_values_df, missing_values_df,  duplicates_df], axis=1)
    
    # Display the summary DataFrame
    display(summary_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Number of plots per row
plots_per_row = 4

for dataset, name in zip(datasets, dataset_names):
    print(f'\n{name}:')
    numeric_columns = dataset.select_dtypes(include=[np.number]).columns
    num_plots = len(numeric_columns)
    
    # If there are no numeric columns, continue to the next dataset
    if num_plots == 0:
        print('No numeric columns in the dataset')
        continue
    
    # Calculate the number of rows needed
    num_rows = num_plots // plots_per_row
    num_rows += num_plots % plots_per_row
    
    # Create a figure and axes with appropriate size
    fig, axs = plt.subplots(num_rows, plots_per_row, figsize=(20, 4*num_rows))
    
    # Flatten the axes array, if there is only one row it is already a 1D array
    if num_rows != 1:
        axs = axs.flatten()

    # Make a boxplot for each numeric column
    for idx, column in enumerate(numeric_columns):
        sns.boxplot(x=dataset[column], ax=axs[idx])
        axs[idx].set_title(f'{column}')
    
    # If there are less plots than subplots, delete the empty subplots
    if num_plots < len(axs):
        for idx in range(num_plots, len(axs)):
            fig.delaxes(axs[idx])

    # Show the plot
    plt.tight_layout()
    plt.show()
```

# Traitement des valeurs manquantes : 

Nous avons maintenant un dataframe unifié qui intègre toutes les informations nécessaires à notre segmentation. Cependant, nous avons plusieurs lignes pour le même client à cause de plusieurs commandes ou plusieurs articles par commande. Pour réaliser une segmentation client, nous devons aggréger ces données pour avoir une seule ligne par client.

Avant cela, nous devrions nettoyer nos données. Cela peut impliquer le traitement des valeurs manquantes et des valeurs aberrantes. Examinons d'abord les valeurs manquantes dans notre dataframe.

ons traiter chacune d'elles :

order_approved_at, order_delivered_carrier_date, order_delivered_customer_date : Ces colonnes sont des timestamps et il est difficile d'imputer les valeurs manquantes de manière significative. Puisque nous ne prévoyons pas d'utiliser ces colonnes directement dans notre clustering, nous pouvons les laisser telles quelles.

review_comment_title, review_comment_message : Ces colonnes contiennent du texte et ne seront probablement pas utilisées dans le clustering. De plus, les valeurs manquantes pourraient simplement indiquer que le client n'a pas laissé de commentaire. Nous pouvons donc laisser ces colonnes telles quelles.

product_category_name, product_name_lenght, product_description_lenght, product_photos_qty : Ces colonnes sont liées au produit et les valeurs manquantes pourraient indiquer que l'information sur le produit est manquante. Nous pourrions remplacer ces valeurs manquantes par une nouvelle catégorie ("Unknown") ou une valeur spécifique (comme 0 ou la médiane).

product_weight_g, product_length_cm, product_height_cm, product_width_cm : Ces colonnes contiennent des informations sur les dimensions du produit. Pour ces colonnes, il pourrait être approprié de remplacer les valeurs manquantes par la médiane ou la moyenne de chaque colonne.

Procédons à l'imputation de ces valeurs manquantes.

## choix et justifications des features pour le modèle de clustering

Plusieurs "customer_unique_id" ont plus d'un "customer_id" associé, ce qui signifie qu'un même client (identifié par "customer_unique_id") a effectué plusieurs transactions (identifiées par "customer_id"). En d'autres termes, "customer_id" est utilisé pour identifier chaque transaction client, tandis que "customer_unique_id" est un identifiant unique pour chaque client.

<b>total_orders, total_payment, mean_order_value, order_size </b>: Ces indicateurs fournissent des informations sur le comportement d'achat des clients, notamment le nombre de commandes qu'ils passent, combien ils dépensent en moyenne, la valeur moyenne de leurs commandes, et la taille de leurs commandes. Ces indicateurs sont souvent utilisés dans les analyses de segmentation des clients, car ils peuvent aider à identifier des groupes de clients ayant des comportements d'achat similaires.

<b>average_review_score, average_payment_installments </b>: Ces indicateurs fournissent des informations sur la satisfaction des clients et leurs préférences de paiement. Le score de révision moyen peut aider à identifier les clients satisfaits et insatisfaits, tandis que le nombre moyen de versements de paiement peut donner une indication des préférences de paiement des clients. Ces indicateurs peuvent également être utiles pour la segmentation des clients.

```python
# Import necessary libraries
import pandas as pd
import numpy as np

# Load datasets (replace with your file paths)
orders = pd.read_csv('olist_orders_dataset.csv')
customers = pd.read_csv('olist_customers_dataset.csv')
items = pd.read_csv('olist_order_items_dataset.csv')
products = pd.read_csv('olist_products_dataset.csv')
payments = pd.read_csv('olist_order_payments_dataset.csv')
reviews = pd.read_csv('olist_order_reviews_dataset.csv')

# Merge datasets
df = pd.merge(orders, customers, on='customer_id')
df = pd.merge(df, items, on='order_id')
df = pd.merge(df, products, on='product_id')
df = pd.merge(df, payments, on='order_id')
df = pd.merge(df, reviews, on='order_id')

# Calculate the number of orders for each customer
order_counts = df.groupby('customer_unique_id')['order_id'].nunique()

# Create a new binary column to indicate if the customer has placed more than one order
df['has_multiple_orders'] = df['customer_unique_id'].map(order_counts > 1).astype(int)

# Calculate the mean order value, order size, and mean review score for each customer
customer_data = df.groupby('customer_unique_id').agg({
    'payment_value': 'mean',
    'order_item_id': 'count',
    'review_score': 'mean',
    'has_multiple_orders': 'max'
}).reset_index()

customer_data.rename(columns={
    'payment_value': 'mean_order_value',
    'order_item_id': 'order_size',
    'review_score': 'mean_review_score',
    'has_multiple_orders': 'has_multiple_orders'
}, inplace=True)

# Determine the most common product category for each customer
most_common_product_category = df.groupby('customer_unique_id')['product_category_name'].agg(
    lambda x: x.value_counts().index[0] if not x.value_counts().empty else np.nan
).reset_index()

most_common_product_category.rename(columns={'product_category_name': 'most_common_product_category'}, inplace=True)

# Merge the dataframes
customer_data = pd.merge(customer_data, most_common_product_category, on='customer_unique_id')

customer_data['most_common_product_category'].fillna('Unknown', inplace=True)

# Display the first few lines of the final dataframe
customer_data.head()
```
